```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Configure numpy and pandas to show exactly 4 decimal places
np.set_printoptions(precision=4, suppress=True)
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
```

```python
import os

# Load the Student Performance Factors dataset (works in both Colab and local Jupyter)
candidate_paths = [
    '/content/Student_Performance_Factors.csv',
    '/content/StudentPerformanceFactors.csv',
    '/content/drive/MyDrive/Student_Performance_Factors.csv',
    'Student_Performance_Factors.csv',
    'StudentPerformanceFactors.csv',
]
file_path = next((p for p in candidate_paths if os.path.exists(p)), 'Student_Performance_Factors.csv')
df = pd.read_csv(file_path)
feature_names = ["Attendance", "Hours_Studied", "Previous_Scores", "Sleep_Hours"]

X_raw = df[feature_names].to_numpy()    # feature matrix, shape (n_samples, n_features)
y = df["Exam_Score"].to_numpy()          # continuous target vector

assert X_raw is not None and y is not None, "Load your dataset above before continuing."
n_samples, n_features = X_raw.shape
print(f"n_samples = {n_samples}, n_features = {n_features}")
print(f"feature_names = {feature_names}")
```

**Q1.** What real-world quantity are you trying to predict, and why is it continuous rather than categorical?

**Q3.** How many samples (n) and how many features (p) does your dataset have?

> *Your answer:* The dataset has $n = 6,607$ samples (students) and $p = 4$ selected numeric features (`Attendance`, `Hours_Studied`, `Previous_Scores`, and `Sleep_Hours`), giving a feature matrix $X_{\text{raw}} \in \mathbb{R}^{6607 \times 4}$ and a target vector $y \in \mathbb{R}^{6607}$ (`Exam_Score`). **Q4.** Before doing any analysis, which features do you expect to be most predictive, and why?
