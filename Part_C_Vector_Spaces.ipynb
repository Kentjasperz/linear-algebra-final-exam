```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
pd.set_option('display.float_format', lambda x: f'{x:.4f}')
```

```python
# Compute the rank of your design matrix X
rank_X = np.linalg.matrix_rank(X)
n_samples, n_features = X.shape
print(f"X is {n_samples} x {n_features}.  rank(X) = {rank_X}")
```

```python
# Compare rank(X) to the number of features.
if rank_X == n_features:
    print(f"rank(X) = {rank_X} = p = {n_features}: X has full column rank, "
          "so the columns (features) are linearly independent.")
else:
    print(f"rank(X) = {rank_X} < p = {n_features}: the columns are linearly dependent "
          "(at least one feature is redundant).")
# Are the columns of X linearly independent? Print a short conclusion.
sv = np.linalg.svd(X, compute_uv=False)
print("Singular values:", sv)
print(f"Condition number: {sv[0] / sv[-1]:.4f}")
```

**Q9.** What is the dimension of the feature space your data lives in (before any reduction)?

> *Your answer:* The feature space is ℝ⁴, so its dimension is p = 4. Each student is a point in 4-dimensional space with coordinates (Attendance, Hours_Studied, Previous_Scores, Sleep_Hours). (The four columns of X, by contrast, are vectors in ℝ⁶⁶⁰⁷, one entry per student.)

**Q10.** What is rank(X)? Is it equal to the number of features p? If not, what does that tell you about your features?

> *Your answer:* `rank(X) = 4`, which equals $p = 4$, so $X$ has full column rank (with singular values $[9138.7708, 1078.6235, 511.1465, 134.6141]$ and condition number $67.8887$). This tells us the four features are linearly independent: none of them can be written exactly as a linear combination of the others, so each one adds information the others do not. As a result, $X^T X$ is invertible and the least-squares coefficients are unique.

**Q11.** In your own words, what is the column space of X, and why does it matter for what predictions your regression model can possibly make?

> *Your answer:* The column space of X is the set of all linear combinations of its four columns, i.e. every vector of the form Xβ = β₁·Attendance + β₂·Hours_Studied + β₃·Previous_Scores + β₄·Sleep_Hours. Since X has rank 4, it is a 4-dimensional subspace sitting inside ℝ⁶⁶⁰⁷. A linear regression's predictions ŷ = Xβ must lie in this subspace, so the model can only produce score vectors that are blends of these four features. The actual exam scores y generally do not lie in it, so the best the model can do is choose the point in the column space closest to y, which is the projection of y onto it.

**Q12.** Describe what a linearly dependent feature would look like in your own dataset (a real or hypothetical example), and why it would be a problem for your model.

> *Your answer:* A hypothetical example: if we added a column Minutes_Studied = 60 × Hours_Studied, it would be an exact multiple of an existing column, so the columns would be linearly dependent and rank(X) would stay 4 while p = 5. The problem is that the extra column adds no new information, XᵀX becomes singular (not invertible), and the normal equations have infinitely many solutions, so the coefficients are not unique. Even if the dependence is only approximate (for example, two features that are almost perfectly correlated), XᵀX becomes ill-conditioned and the coefficients become unstable and hard to interpret.
